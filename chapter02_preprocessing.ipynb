# 1. Handling Missing Data

# A. SimpleImputer()
  - Strategi: Mengganti dengan rata-rata (mean), median, atau modus.
  - Kelebihan: Sederhana, cepat; baik jika persentase missing $< 5\%$

# B. KNNImputer()
  - Strategi: Menggunakan $k$-tetangga terdekat berbasis jarak Euclidean
  - Kelebihan: Mengamankan struktur lokal data numerik

# C. IterativeImputer()
  - Strategi: Memodelkan setiap fitur sebagai fungsi regresi dari fitur lain
  - Kelebihan: Menjaga hubungan antar-variabel/multivariat



```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

# 1. Membuat dataset sintetis dengan missing values (~20%)
np.random.seed(2024)
n_samples, n_features = 20, 4
data = {f"Feature{i+1}": np.random.uniform(0, 100, n_samples) for i in range(n_features)}
df = pd.DataFrame(data)

# Memasukkan NaN secara acak
for col in df.columns:
    mask = np.random.random(n_samples) < 0.2
    df.loc[mask, col] = np.nan

# 2. Imputasi dengan SimpleImputer (Mean)
simple_imp = SimpleImputer(strategy="mean")
df_simple = pd.DataFrame(simple_imp.fit_transform(df), columns=df.columns)

# 3. Imputasi dengan KNNImputer (k=2)
knn_imp = KNNImputer(n_neighbors=2)
df_knn = pd.DataFrame(knn_imp.fit_transform(df), columns=df.columns)

# 4. Imputasi dengan IterativeImputer
iter_imp = IterativeImputer(random_state=2024, max_iter=10)
df_iter = pd.DataFrame(iter_imp.fit_transform(df), columns=df.columns)

print("--- Hasil Perbandingan Nilai Imputasi Feature1 ---")
f1_nan_idx = df[df['Feature1'].isna()].index.tolist()
print(f"Indeks NaN pada Feature1 : {f1_nan_idx}")
print(f"SimpleImputer (Mean)     : {[round(df_simple.loc[i, 'Feature1'], 2) for i in f1_nan_idx]}")
print(f"KNNImputer (k=2)         : {[round(df_knn.loc[i, 'Feature1'], 2) for i in f1_nan_idx]}")
print(f"IterativeImputer         : {[round(df_iter.loc[i, 'Feature1'], 2) for i in f1_nan_idx]}")
```

- SimpleImputer(strategy="mean"): Mengisi seluruh NaN pada suatu kolom dengan nilai rata-rata dari data yang teramati di kolom tersebut. Metode ini cepat tetapi dapat mengurangi varians asli data.

- KNNImputer(n_neighbors=2): Mencari 2 sampel terdekat yang memiliki fitur serupa, lalu mengisi NaN berdasarkan rata-rata terbobot dari tetangga tersebut.

- IterativeImputer(): Menggunakan pendekatan mirip MICE (Multiple Imputation by Chained Equations),Berikut adalah sajian lengkap untuk Chapter 2: Pre-Model Workflow and Data Preprocessing dengan format terstruktur: Teori -> Kode -> Penjelasan Kode untuk setiap topik bahasannya.

## Handling Missing Data

Data yang hilang merupakan masalah paling umum dalam dataset dunia nyata. Sebagian besar algoritma scikit-learn akan menghasilkan error jika terdapat nilai NaN. Terdapat tiga strategi utama imputasi:

1. SimpleImputer(): Mengganti nilai hilang menggunakan statistik sederhana (rerata, median, atau modus). Cepat dan deterministik, ideal untuk data hilang $<5\%$.

2. KNNImputer(): Mengisi nilai hilang berdasarkan sampel terdekat ($k$-nearest neighbors) menggunakan jarak Euclidean. Menjaga struktur lokal data.

3. IterativeImputer(): Memodelkan setiap fitur yang hilang sebagai fungsi regresi dari fitur lainnya secara berulang (multivariate).

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

# Membuat dummy data dengan ~20% missing values
np.random.seed(2024)
data = {f"Feature{i+1}": np.random.uniform(0, 100, 20) for i in range(10)}
df = pd.DataFrame(data)

for col in df.columns:
    mask = np.random.random(20) < 0.2
    df.loc[mask, col] = np.nan

# 1. SimpleImputer (Mean)
simple_imp = SimpleImputer(strategy="mean")
df_simple = pd.DataFrame(simple_imp.fit_transform(df), columns=df.columns)

# 2. KNNImputer (k=2)
knn_imp = KNNImputer(n_neighbors=2)
df_knn = pd.DataFrame(knn_imp.fit_transform(df), columns=df.columns)

# 3. IterativeImputer
iter_imp = IterativeImputer(random_state=2024, max_iter=10)
df_iter = pd.DataFrame(iter_imp.fit_transform(df), columns=df.columns)

f1_missing = df[df['Feature1'].isna()].index.tolist()
print(f"Indeks missing pada Feature1: {f1_missing}")
if f1_missing:
    print(f"SimpleImputer (mean)   : {df_simple.loc[f1_missing[0], 'Feature1']:.2f}")
    print(f"KNNImputer (k=2)       : {df_knn.loc[f1_missing[0], 'Feature1']:.2f}")
    print(f"IterativeImputer       : {df_iter.loc[f1_missing[0], 'Feature1']:.2f}")
```

- SimpleImputer(strategy="mean"): Menghitung rerata dari nilai yang ada pada Feature1 lalu memasukkannya ke semua sel NaN di kolom tersebut.

- KNNImputer(n_neighbors=2): Mencari 2 baris data terdekat yang memiliki fitur serupa, kemudian mengimputasi nilai hilang menggunakan rerata dari tetangga tersebut. Nilai imputasi bervariasi tergantung sampelnya.

- enable_iterative_imputer: Wajib diimpor karena IterativeImputer masih berstatus experimental di scikit-learn.

### SimpleImputer -- Fast Statistical Replacement

`SimpleImputer()` replaces each missing value with a summary statistic computed from the non-missing values in the same feature column. The available strategies are:

- `"mean"` -- Replace with the column mean $\mu_j = \frac{1}{n_{\text{obs}}} \sum_{i: x_{ij} \neq \text{NaN}} x_{ij}$
- `"median"` -- Replace with the column median (robust to outliers)
- `"most_frequent"` -- Replace with the mode (useful for categorical features)
- `"constant"` -- Replace with a user-specified fill value

The **mean** strategy is the default and most common choice for numerical features when the data is approximately symmetric (no heavy outliers).

```python
from sklearn.impute import SimpleImputer

# Initialize with mean strategy
imputer = SimpleImputer(strategy="mean")

# Fit (learn column means) and transform (fill NaNs)
imputed_data = imputer.fit_transform(df)
imputed_df = pd.DataFrame(imputed_data, columns=df.columns)

print("Learned means for each feature:")
for feat, mean_val in zip(df.columns, imputer.statistics_):
    print(f"  {feat}: {mean_val:.4f}")
print()
print("Missing values after imputation:", int(imputed_df.isna().sum().sum()))
print()
print("First 6 rows after SimpleImputer (mean):")
print(imputed_df.head(6).to_string())
```

## Scaling Techniques

Fitur dengan skala berbeda dapat mengganggu algoritma berbasis jarak (KNN, SVM) dan pemutakhiran gradien.

- StandardScaler(): Mentransformasi data agar berpusat di rerata $\mu = 0$ dengan varians $\sigma^2 = 1$ ($z = \frac{x - \mu}{\sigma}$). Cocok untuk data berdistribusi normal.
- MinMaxScaler(): Mengubah rentang data menjadi $[0, 1]$ ($x' = \frac{x - x_{\min}}{x_{\max} - x_{\min}}$). Sensitif terhadap outlier.
- Normalizer(): Mentransformasi data secara baris/per sampel (bukan kolom/fitur) agar memiliki unit norm ($L1$ atau $L2$). Ideal untuk data teks/TF-IDF.

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, Normalizer

# Menggunakan data terimputasi dari langkah sebelumnya
data_to_scale = df_iter.copy()

# 1. StandardScaler (Z-score)
std_scaler = StandardScaler()
df_std = pd.DataFrame(std_scaler.fit_transform(data_to_scale), columns=df.columns)

# 2. MinMaxScaler (Range 0 - 1)
minmax_scaler = MinMaxScaler()
df_minmax = pd.DataFrame(minmax_scaler.fit_transform(data_to_scale), columns=df.columns)

# 3. Normalizer (L2 Norm per baris)
norm_scaler = Normalizer(norm='l2')
df_norm = pd.DataFrame(norm_scaler.fit_transform(data_to_scale), columns=df.columns)

print("StandardScaler - Column Means:", np.round(df_std.mean().values[:3], 4))
print("StandardScaler - Column Stds :", np.round(df_std.std().values[:3], 4))
print("\nMinMaxScaler - Min Values   :", np.round(df_minmax.min().values[:3], 4))
print("MinMaxScaler - Max Values   :", np.round(df_minmax.max().values[:3], 4))
print("\nNormalizer - Row 0 L2 Norm   :", np.sqrt((df_norm.iloc[0, :]**2).sum()))
```

- df_std: Menghasilkan fitur yang memiliki rerata $0$ dan deviasi standar $1$.

- df_minmax: Mengompresi seluruh nilai fitur ke dalam batas minimum $0.0$ dan maksimum $1.0$.

- df_norm: Menghitung norma $L2$ pada setiap baris sehingga vektor tiap sampel berada pada permukaan hypersphere bernilai $1.0$.

## Categorical Encoding & ColumnTransformer

Data kategorikal harus diubah menjadi angka sebelum dimasukkan ke dalam model.

- Nominal (tanpa urutan, misal: lokasi): Gunakan OneHotEncoder() untuk membuat kolom biner ($0/1$) baru.
- Ordinal (memiliki tingkatan, misal: jabatan): Gunakan LabelEncoder() / OrdinalEncoder().
- ColumnTransformer(): Memungkinkan penerapan transformasi berbeda pada kolom numerik dan kategorikal secara bersamaan dalam satu langkah.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Data campuran numerik & kategorikal
mixed_df = pd.DataFrame({
    "Age": [25, 45, 35, 50],
    "Salary": [50000, 80000, 62000, 110000],
    "Department": ["IT", "HR", "Sales", "IT"]
})

# Menentukan transformasi berbasis kolom
ct = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), ["Age", "Salary"]),
        ("cat", OneHotEncoder(sparse_output=False), ["Department"])
    ]
)

transformed_data = ct.fit_transform(mixed_df)
cat_cols = ct.named_transformers_["cat"].get_feature_names_out(["Department"])
all_cols = ["Age_scaled", "Salary_scaled"] + list(cat_cols)

df_transformed = pd.DataFrame(transformed_data, columns=all_cols)
print(df_transformed.to_string())
```

- ColumnTransformer: Menerapkan StandardScaler khusus pada kolom Age dan Salary, serta OneHotEncoder pada kolom Department.

- et_feature_names_out(): Mengambil nama kolom biner baru yang dihasilkan oleh OneHotEncoder (seperti Department_HR, Department_IT, Department_Sales).

## Building Pipelines to Prevent Data Leakage

Pipeline() mengaitkan beberapa tahap pra-pemrosesan dan estimator dalam satu objek.

Aturan Utama: Lakukan pemisahan dataset (train_test_split) sebelum memasukkannya ke dalam pipeline. Pipeline menjamin bahwa parameter transformasi (seperti rerata imputer atau standar deviasi scaler) hanya dipelajari dari data pelatihan melalui fit_transform(), kemudian diterapkan ke data uji menggunakan transform() untuk mencegah kebocoran data (data leakage).

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

X = df_transformed.iloc[:, :-1]
y = [0, 1, 0, 1]  # Dummy target

# Split data terlebih dahulu
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)

# Membuat pipeline
prep_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])

# Fit & transform pada data latih
X_train_prep = prep_pipeline.fit_transform(X_train)

# Hanya transform pada data uji
X_test_prep = prep_pipeline.transform(X_test)

print("Langkah Pipeline:", [name for name, _ in prep_pipeline.steps])
print("Shape Data Latih Teraplikasi:", X_train_prep.shape)
print("Shape Data Uji Teraplikasi  :", X_test_prep.shape)
```

- prep_pipeline.fit_transform(X_train): Mempelajari statistik data latih dan langsung mentransformasikannya.

- prep_pipeline.transform(X_test): Mengaplikasikan statistik yang dipelajari dari X_train ke X_test tanpa melakukan kalkulasi ulang pada data uji, menjamin tidak ada informasi data uji yang bocor.


## Feature Engineering (Extraction & Selection)

Feature Engineering mencakup dua aktivitas:

1. Ekstraksi Fitur: Membuat fitur baru dari fitur yang ada.
  - PolynomialFeatures(): Menghasilkan interaksi perkalian dan derajat polinomial fitur ($x_1^2, x_1 x_2$).
  - KBinsDiscretizer(): Mengubah variabel kontinu menjadi kategori/bin interval.

2. Seleksi Fitur: Memilih subset fitur terbaik untuk mengurangi noise dan kompleksitas.
  - RFE() (Recursive Feature Elimination): Eleminasi fitur kurang penting secara iteratif mundur.
  - SelectFromModel(): Memilih fitur berdasarkan batas ambang (threshold) nilai koefisien atau tingkat kepentingan dari satu kali fitting.

```python
from sklearn.preprocessing import PolynomialFeatures, KBinsDiscretizer
from sklearn.feature_selection import RFE, SelectFromModel
from sklearn.linear_model import LinearRegression

# A. Extraction: PolynomialFeatures
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X_train_prep)

# B. Extraction: KBinsDiscretizer
kbins = KBinsDiscretizer(n_bins=3, encode="ordinal", strategy="uniform")
X_binned = kbins.fit_transform(X_train_prep)

# C. Selection: RFE
rfe = RFE(estimator=LinearRegression(), n_features_to_select=2)
rfe.fit(X_poly, y_train)

# D. Selection: SelectFromModel
sfm = SelectFromModel(estimator=LinearRegression(), threshold="mean")
sfm.fit(X_poly, y_train)

print(f"Jumlah Fitur Asli         : {X_train_prep.shape[1]}")
print(f"Fitur Setelah Polynomial  : {X_poly.shape[1]}")
print(f"Fitur Dipilih via RFE     : {rfe.support_.sum()}")
print(f"Fitur Dipilih via SFM     : {sfm.transform(X_poly).shape[1]}")
```

- PolynomialFeatures(degree=2): Menambah ekspansi fitur termasuk kombinasi pasangan perkalian antar fitur.

- KBinsDiscretizer: Membagi nilai kontinu menjadi 3 rentang bin diskret ($0, 1, 2$).

- RFE: Membuang fitur paling tidak penting satu per satu hingga tersisa 2 fitur terbaik.

- SelectFromModel: Menyaring dan menyisakan fitur yang memiliki bobot koefisien di atas nilai rerata.
